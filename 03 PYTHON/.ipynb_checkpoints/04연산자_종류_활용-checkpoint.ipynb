# 논리연산자
* and, or, not

* and는 두 값이 모두 참이어야 참

```python
True and True
```

```python
True and False
```

```python
False and False
```

or 두 값중 하나라도 참이면 참, 두 값 모두가 거짓이어야 거짓

```python
True or True
```

```python
True or False
```

```python
False or True
```

```python
False or False
```

not은 값을 반대로 뒤집음 True => False, False => True

```python
not True
```

```python
not False
```

* and, or, not의 우선순위
* not, and, or순으로 판단

```python
not True and False or not False
```

```python
10 == 10 and 10 != 5
```

# 식별연산자
* is, is not (object, data type)이 같은지 비교

```python
print(a)
print(b)
print(c)
```

```python
print(a == b) # a는 정수, b는 실수 3.0이라 값이 같음
print(a is b) # a는 정수, b는 실수 3.0이라 데이터 타입이 다름
```

```python
d = 3
```

```python
print( a is d) # a와 d가 모두 int 3이기 때문에 True
```

```python
print(a is not b)
```
