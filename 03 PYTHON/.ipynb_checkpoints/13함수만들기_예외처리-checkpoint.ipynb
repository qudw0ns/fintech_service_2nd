* 숫자를 3개 입력받아서 첫번째 숫자는 시작, 두번째 숫자는 끝, 마지막 숫자는 step으로 작동하게 하세요.
* 먼저 숫자를 생성한 결과를 출력하고
* 숫자리스트에 * 10 한 결과를 다시 출력하세요.
* num_maker(1, 11, 2)
* 결과 1,3,5,7,9

```python
def num_maker(a, b, c):
    result = list(range(a, b, c))
    result2 = []
    print(result)
    for i in result:
        result2.append(i*10)
    print(result2)
    return result, result2
```

```python
num_maker(1, 11, 2)
```

# 중간에 함수를 종료시키기 위한 용도의 return
* 반복문에서 break 처럼 활용 가능
* 함수 내에서 return이 실행되면 함수는 종료

* 1에서 100까지 숫자 중 짝수만 더해서 출력하세요.
* 단, 숫자가 50이상이 되면 멈추시오
* 더해지는 과정도 출력

```python
def even(num):
    hap = 0
    for i in num:
        if i < 50:
            if i % 2 == 0:
                hap += i
                print(hap, end=" ")
        else:
            break
    return hap
```

```python
even(list(range(1, 100)))
```

```python
def even2(num):
    hap = 0
    for i in num:
        if i < 50:
            if i % 2 == 0:
                print(f"{i}+{hap}=", end="")
                hap += i
                print(f"{hap}", end=" ")
        else:
            return hap
```

```python
even2(list(range(1, 100)))
```

* 점수를 입력받아서 91-100이면 A학점, 81-90 B학점, 71-80 C학점, 61-70 D학점, 60 이하는 F학점을 출력하는 student_grade 함수를 만드세요.
* 점수를 입력받음 input()
* student_grade 함수에 입력받은 값을 넣어줌
* student_grade 함수에서는 입력받은 값을 비교해서 학점 출력

```python
def student_grade(score):
    if 0 <= score <= 100:
        if 91 <= score <= 100:
            return "A학점"
        elif 81 <= score <= 90:
            return "B학점"
        elif 71 <= score <= 80:
            return "C학점"
        elif 61 <= score <= 70:
            return "D학점"
        else:
            return "F학점"
```

```python
num = int(input("점수을 입력하세요"))
result = student_grade(num)
print(result)
```

* my_sum
* 여러개의 숫자를 입력받으면 합계를 구해주는 함수

```python
def my_sum(*nums):
    hap = 0
    for i in nums:
        hap += i
    return hap
```

* 위치인수와 가변인수 동시에 받기
* 위치인수와 가변인수를 동시에 받을 때는 반드시 위치인수가 먼저 와야 한다.
* def muls(a, * args): OK

```python
# 정수와 리스트를 입력받아 리스트의 요소에 입력받은 정수를 곱하는 함수
```

```python
def muls(a, *args):
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
l2 = [1,2,3,4,5]
```

```python
muls(2, *l2, *l2)
```

```python
def muls(*args, a):
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
muls(*l2, 3)
```

#### 매개변수 = 함수에서 자료를 받을 때 사용하는 변수
* 위치인수(a): 위치가 고정된 매개변수, 위치인수는 키워드 인수로 사용가능(변수명=값)
* 가변인수(*args): 여러 개의 자료를 동시에 받을 수 있음. => tuple로 변환되어 들어옴
* 키워드 가변인수(**kwargs): 여러개의 자료를 동시에 받을 수 있는데 key=value 형태로 받는 인수 => dict로 들어옴


* 매개변수에 초기값 지정하기
* 초기값이 지정된 매개변수는 반드시 마지막에 위치해야 함
* 위치인수, 키워드 인수, 가변인수, 키워드 가변인수를 동시에 다 사용할 경우
* 위치인수 => 키워드 인수(기본값), 가변인수, 키워드 가변인수 순서로 사용한다.

```python
def personal_info6(idx, address="비공개", *args, **kwargs):
    print(f"idx: {idx}")
    print(f"address: {address}")
    print(f"*args: {args}")
    print(f"**kwargs: {kwargs}")
```

* 1에서 10까지의 수 중에서 3의 배수만 문자열로 바꾸기

```python
def tostr(x):
    result=[]
    for i in x:
        if i % 3 == 0:
            result.append(str(i))
        else:
            result.append(i)
    return result
```

```python
result_str = tostr(list(range(1, 11)))
print(result_str)
```
