# dict 함수를 사용해서 dict 만들기

* dict(key1=value1, key2=value2, key3=value3)
* key에 ', ""를 사용하지 않는다.
* key는 모두 문자형으로 변환된다.
* value가 문자형 자료면 반드시 ""가 필요하다.

```python
dict_ex1=dict(정수=100, 실수=3.5, 문자="문자열", 리스트=[1,2,3], 튜플=(1,2,3), 딕셔너리={"key", "vale"})
```

```python
dict_ex1
```

* dict([(key1, value1), (key2, value2)]
* 리스트 안에 key, value 쌍으로 된 튜플을 넣어서 만듬

```python
dict_ex2 = dict([('정수',3), ('실수', 3.5), ('문자열', '문자열입니다.'),("list", [1,2,3]),("tuple", (1,2,3)), ("dict", {"key" : "value"})])
```

```python
dict_ex2
```

* dict(zip([key1, key2], [value1, value2]))
* key 묶음 리스트와 value 묶음 리스트를 zip 함수로 압친 후 dict 생성
* key 묶음의 요소 개수와 value 묶음의 요소 개수가 일치해야함

```python
for a in zip([1,2,3,4], [5,6,7,8]):
    print(a)
```

```python
dict_ex3 = dict(zip(['정수', '실수', '문자열', '리스트', '튜플', "딕셔너리"], 
                    [10, 13.5, "문자열입니다.", [123], (123), {10:30}]))
dict_ex3
```

## 딕셔너리에서 key, value, (key, value)를 가져오는 함수
* keys(): 딕셔너리에서 key만 출력
* values(): 딕셔너리에서 value만 출력
* items(): 딕셔너리의 key와 value를 튜플 묶음 형태로 출력(key, value)

* keys() : 딕셔너리에서 key만 출력
* list() 함수로 형변환 해주면  list처럼 사용 가능

```python
data.keys()
```

```python
type(data.keys())
```

```python
type(list(data.keys()))
```

```python
list(data.keys())[1]
```

```python
for i in data.keys():
    print(i)
```

* values(): 딕셔너리에서 value만 추출
* list()로 형변환을 해주면 list처럼 사용가능
* 반복문에서 사용가능

```python
data.values()
```

```python
list(data.values())[0][0]['name']
```

```python
for i in data.values():
    print(i)
```

```python
for i in data:
    print(i)
```

* items(): key, value를 튜플 형태의 쌍으로 출력
* list() 형변환을 통해서 list처럼 사용 가능

```python
d.items()
```

```python
list(d.items())[0:2]
```

```python
for item in d.items():
    print(item, end="\n\n")
```
