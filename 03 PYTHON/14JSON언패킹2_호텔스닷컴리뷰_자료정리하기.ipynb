```python
import json
import pandas as pd
```

```python
with open("./data/hotels_korea_reviews_text.json", "r", encoding="utf-8") as file:
    data = file.read()
    ho = json.loads(data)
```

```python
len(ho)
```

# 호텔 인포 테이블

```python
hi = []
nm = []
sr = []
rv = []
rt = []
cl = []
sv = []
co = []
cd = []
nh = []
rg = []
lo = []
```

```python
for i in range(len(ho)):
    hotelid = ho[i]["data"]["common"]["injected_data"]["commonDataBlock"]["property"]["hotelId"]
    name = ho[i]["data"]["body"]["propertyDescription"]["name"]
    starRating = ho[i]["data"]["body"]["propertyDescription"]["starRating"]
    reviewsTotal = ho[i]["data"]["body"]["reviewContent"]["pagination"]["reviewsTotal"]
    rating = ho[i]["data"]["body"]["reviewContent"]["overall"]["rating"]
    cleanliness = ho[i]["data"]["body"]["reviewContent"]["overall"]["ratingAspects"].get("cleanliness", "none")
    service = ho[i]["data"]["body"]["reviewContent"]["overall"]["ratingAspects"].get("service", "none")
    comfort = ho[i]["data"]["body"]["reviewContent"]["overall"]["ratingAspects"].get("comfort", "none")
    condition = ho[i]["data"]["body"]["reviewContent"]["overall"]["ratingAspects"].get("condition", "none")
    neighbourhood = ho[i]["data"]["body"]["reviewContent"]["overall"]["ratingAspects"].get("neighbourhood", "none")
    locality = ho[i]["data"]["body"]["propertyDescription"]["address"]["locality"]
    region = ho[i]["data"]["body"]["propertyDescription"]["address"].get("region", "locality")
    
    
    hi.append(hotelid)
    nm.append(name)
    sr.append(starRating)
    rv.append(reviewsTotal)
    rt.append(rating)
    cl.append(cleanliness)
    sv.append(service)
    co.append(comfort)
    cd.append(condition)
    nh.append(neighbourhood)
    lo.append(locality)
    rg.append(region)
```

```python
result = dict(Hotelid=hi, 호텔이름=nm, 호텔등급=sr, 총리뷰수=rv, 평점=rt, 청결도=cl, 서비스=sv, 편안함=co, 시설상태=cd, 주변환경=nh, 시도=lo, 지역=rg)
```

```python
result_df = pd.DataFrame(result)
result_df
```

# 사용자 테이블

```python
hi2 = []
rd2 = []
it2 = []
tt2 = []
ra2 = []
de2 = []
```

```python
for i in range(len(ho)):
    for h in range(len(ho[i]["data"]["body"]["reviewContent"]["reviews"]["hermes"]["groups"])):
        for g in range(len(ho[i]["data"]["body"]["reviewContent"]["reviews"]["hermes"]["groups"][h]["items"])):
            hotelid2 = ho[i]["data"]["common"]["injected_data"]["commonDataBlock"]["property"]["hotelId"]
            reviewDate = ho[i]["data"]["body"]["reviewContent"]["reviews"]["hermes"]["groups"][h]["items"][g]["reviewDate"]
            itineraryId = ho[i]["data"]["body"]["reviewContent"]["reviews"]["hermes"]["groups"][h]["items"][g].get("itineraryId", "none")
            tripType = ho[i]["data"]["body"]["reviewContent"]["reviews"]["hermes"]["groups"][h]["items"][g]['tripType']
            rating2 = ho[i]["data"]["body"]["reviewContent"]["reviews"]["hermes"]["groups"][h]["items"][g]['rating']
            description = ho[i]["data"]["body"]["reviewContent"]["reviews"]["hermes"]["groups"][h]["items"][g]["description"]

            hi2.append(hotelid2)
            rd2.append(reviewDate)
            it2.append(itineraryId)
            tt2.append(tripType)
            ra2.append(rating2)
            de2.append(description)
```

```python
result2 = dict(Hotelid2=hi2, 리뷰일=rd2, 고객id=it2, 여행타입=tt2, 평점=ra2, 리뷰=de2)
```

```python
result2_df = pd.DataFrame(result2)
result2_df
```

```python
result2_df.to_csv("호텔스닷컴리뷰.csv", encoding="utf-8-sig", index=False)
```
