클래스에 기능 추가하기

````python
class FourCal():
    # 데이터를 입력받는 역할의 setdata method
    def setdata(self, num1, num2):
        self.num1 = num1
        self.num2 = num2
        print("self.num1", self.num1)
        print("self.num2", self.num2)
        
    #덧셈
    def add(self):
        result = self.num1 + self.num2
        return result
    
    #뺄샘
    def sub(self):
        result = self.num1 - self.num2
        return result
    
    #곱셈
    def mul(self):
        result = self.num1 * self.num2
        return result
    
    #나눗셈
    def div(self):
        result = self.num1 / self.num2
        return result
````

````python
sam = FourCal()
````

````python
sam.sub()
````

````python
sam.mul()
````

````python
sam.div()
````

# 생성자
* 클래스를 실행해서 인스턴스 턴스를 만들 때 초기값을 받도록 해주는 메소드
* 생성자(constructor)는 객체(인스턴스)가 생성될 때 자동으로 호출되는 메소드, 주로 초기값을 받을 때 사용
* 클래스 내에서 메소드명으로 \_\_init\_\_를 사용하면 그것이 생성자가 된다.

* 생성자를 이용해 초기값을 입력받는 FourCal2

````python
class FourCal2():
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2
    def add(self):
        result = self.num1 + self.num2
        return result
    def sub(self):
        result = self.num1 - self.num2
        return result
    def mul(self):
        result = self.num1 * self.num2
        return result
    def div(self):
        result = self.num1 / self.num2
        return result
````

````python
gildong.add()
````

* 부모 클래스에서 상속한 메소드를 자식 클래스에서 동일한 이름으로 재정의

* FourCal2는 num2에 0이 입력될 경우 div메서드 실행 시 0으로 나누게 되기 때문에 에러발생
* SafeFourCal을 만들고 FourCal2fmf 상속받은 후 div 함수를 재정의 하여 에러를 해결

````python
class SafeFourCal(FourCal2):
    def div(self):
        if self.num2 == 0:
            return 0
        else:
            return self.num1 / self.num2
````

# 메소드 오버로딩
* 한 클래스 안에서 동일 메소드의 기능을 다르게 재정의
* 동일한 이름의 메소드가 기능이 약간 변경된 채로 여러 개 존재
* def add(num1, num2)
* def add(num1, num2, num3)
* def add(num1, num2, num3, num4)

````python
class OverloadingEx():
    def add(self, x, y):
        return x + y
    def add(self, x, y, z):
        return x + y + z
    def add(self, x, y, z, a):
        return x + y + z + a
````

````python
sam = OverloadingEx()
````

````python
sam.add(10, 20, 30, 40)
````

````python
class OverloadingEx2():
    @dispatch(int, int)
    def add(self, x, y):
        return x + y
    @dispatch(int, int, int)
    def add(self, x, y, z):
        return x + y + z
    @dispatch(int, int, int, int)
    def add(self, x, y, z, a):
        return x + y + z - a
    @dispatch(float, float, float, float)
    def add(self, x, y, z, a):
        return (x + y) - (z * a)
````

# 정적메소드 static method
* 인스턴스를 만들지 않고도 클래스의 메소드를 바로 사용할 수 있는 메소드
* 메소드 위에 @staticmethod라는 데코레이터를 붙여서 만든다.
* static method는 self를 받지 않으므로 인스턴스 속성, 인스턴스 메소드가 필요하지 않을 때 사용

````python
class Cal():
    @staticmethod
    def add(a, b):
        print(a+b)
        
    @staticmethod
    def sub(a, b):
        print(a-b)
````

# 클래스 메서드 class method
* 메소드 위에 @classmethod를 붙여서 만든다.
* 클래스 메소드의 첫 번째 매개변수에는 cls를 지정해야 한다.
* 클래스 메소드도 staticmethod처럼 인스턴스 생성 없이 호출이 가능
* 클래스 메소드는 메소드 안에서 클래스 속성, 클래스 메소드에 접근해야 할 때 사용한다.

````python
class Factory():
    n_coffee_machine = 0 # 클래스 속성
    
    def __init__(self):
        Factory.n_coffee_machine += 1 # 인스터스가 생성될 떄 클래스 속성에 1을 더함
        
    @classmethod
    def print_n_coffee_machine(cls):
        print(f"{cls.n_coffee_machine}개가 제조되었습니다.")
````

````python
Factory.print_n_coffee_machine()
````

````python
sam = Factory()
````

````python
sally = Factory()
````

# 추상 클래스 abc(abstract base class)
* abc 모듈 필요
* 추상 클래스는 추상 클래스를 상속 받아 클래스를 만들 때 반드시 만들어야 하는 메소드를 지정해주는 클래스
* 인터페이스
* 규칙을 부여해서 그 규칙에 따라 클래스를 만들도록 강제하는 역할
* 추상 클래스를 상속받은 클래스는 반드시 추상 클래스에서 정의한 메소드를 overriding으로 구현해야 한다.
* 이력서 양식에 맞추어 내용 적기

````python
from abc import *
````

````python
# 추상 클래스
class CoffeeBase(metaclass=ABCMeta):
    @abstractmethod
    def bean_input(self):
        pass
    
    @abstractmethod
    def grind(self):
        pass
````

````python
# 추상 클래스는 인스턴스 생성 불가
sam = CoffeeBase()
````

````python
class CoffeeMachine(CoffeeBase):
    def bean_input(self):
        print('커피 원두 넣기')
````

````python
sam = CoffeeMachine()
````

````python
class CoffeeMachine2(CoffeeBase):
    def bean_input(self):
        print('커피 원두 넣기')
        
    def grind(self):
        print("커피 원두 갈기")
````

````python
sam = CoffeeMachine2
````

````python
sam.bean_input()
````

````python
sam.grind()
````
