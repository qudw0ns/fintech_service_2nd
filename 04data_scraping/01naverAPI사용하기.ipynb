```python
import json
```

```python
data = json.loads(data)
```

```python
data.keys()
```

```python
def text_clean(x):
    result = x.replace("<b>","")
    return result
```

```python
result = {}
for item in data["items"]:
    for key, value in item.items():
        value = text_clean(value)
        result.setdefault(key, []).append(value)
result
```
