# 수집한 자료 반복문으로 모으기

```python
result = []
for i in range(1, 44):
    url = "http://apis.data.go.kr/B553077/api/open/sdsc2/storeListInUpjong"
    payload = {'servicekey': '7V78PQCO+ycH8PsNQjW3Qt5lJ/b+7v6wSXIWFgRZEZB9w2PIR97FSFP/Gfy/v3MroRk2J7JykRUvaxOkpiQszQ==',
                'divId': 'indsSclsCd', "key":"I21006", "numOfRows": 1000, "pageNo": i, "type":"json"}
    r = requests.get(url, params=payload)
    data = r.json()
    result.append(data)
```

```python
data["body"]["numOfRows"] * data["body"]["pageNo"] 
```

```python
result[0]
```

```python
result3 = {}
for page in result:
    for item in page["body"]["items"]:
        for key, value in zip(page["header"]["columns"], item.values()):
            result3.setdefault(key, []).append(value)
```

```python
df = pd.DataFrame(result3)
df
```

```python
df.to_csv("공공데이터_상권정보_202504.csv", encoding="utf-8", index=False)
```
