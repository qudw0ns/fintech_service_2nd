# 2. Selenium을 사용한 동적 웹페이지 스크래핑

# 4. 에러 처리와 함께 사용하는 예제

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from time import sleep

def scrape_website(url):
    try:
        # 요청에 User-Agent 추가
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
        }
        
        response = requests.get(url, headers=headers)
        response.raise_for_status()  # 에러 체크
        
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # 데이터 추출 로직
        data = []
        elements = soup.find_all('div', class_='item')
        
        for element in elements:
            title = element.find('h2').text.strip()
            price = element.find('span', class_='price').text.strip()
            data.append({'Title': title, 'Price': price})
            
        return data
            
    except requests.RequestException as e:
        print(f"Error fetching the URL: {e}")
        return None
    except Exception as e:
        print(f"Error parsing the data: {e}")
        return None

# 여러 페이지 스크래핑
urls = ['https://example.com/page1', 'https://example.com/page2']
all_data = []

for url in urls:
    data = scrape_website(url)
    if data:
        all_data.extend(data)
    sleep(1)  # 웹사이트 부하 방지를 위한 지연

# 결과 저장
if all_data:
    df = pd.DataFrame(all_data)
    df.to_csv('scraped_data.csv', index=False)
```
