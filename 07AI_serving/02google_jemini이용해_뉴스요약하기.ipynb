```python
import gradio as gr
from google import genai

# ✅ API 키 설정
apikey = apikey  # 여기에 본인의 Gemini API 키 입력
client = genai.Client(api_key=apikey)

# ✅ Gemini 응답 함수
def ask_gemini(prompt):
    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=prompt,
        )
        return response.text
    except Exception as e:
        return f"⚠️ 오류 발생: {str(e)}"

# ✅ Gradio UI 구성
with gr.Blocks() as demo:
    gr.Markdown("## 🌟 Gemini 2.0 Flash 챗봇")

    with gr.Row():
        with gr.Column():
            input_box = gr.Textbox(lines=4, placeholder="질문을 입력하세요", label="🧠 질문")
            submit_btn = gr.Button("Gemini에게 물어보기")

        with gr.Column():
            output_box = gr.Textbox(lines=10, label="🔍 Gemini의 응답", interactive=False)

    submit_btn.click(fn=ask_gemini, inputs=input_box, outputs=output_box)

demo.launch()
```

```python
import os
import gradio as gr
from dotenv import load_dotenv
import requests
import pandas as pd
from bs4 import BeautifulSoup
from google import genai

# 🌐 환경변수에서 Gemini API 키 로드
load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

# 🔑 네이버 오픈 API 인증 정보
NAVER_CLIENT_ID = "sk2iKdMANqpvG3Pa2WIJ"
NAVER_CLIENT_SECRET = "by6xHZcrlB"

# 🔍 네이버 뉴스 검색 함수
def search_news(query):
    url = "https://openapi.naver.com/v1/search/news.json"
    headers = {
        "X-Naver-Client-Id": NAVER_CLIENT_ID,
        "X-Naver-Client-Secret": NAVER_CLIENT_SECRET
    }
    params = {
        "query": query,
        "display": 100,
        "start": 1,
        "sort": "date"
    }
    res = requests.get(url, headers=headers, params=params)
    items = res.json().get("items", [])
    df = pd.DataFrame(items)
    
    if df.empty:
        return pd.DataFrame(columns=["title_clean", "originallink"])

    # HTML 태그 제거 및 날짜 정렬
    df['title_clean'] = df['title'].str.replace(r"<.*?>", "", regex=True)
    df['pubDate'] = pd.to_datetime(df['pubDate'], errors='coerce')
    df = df.sort_values(by='pubDate', ascending=False)
    return df[['title_clean', 'originallink']].head(10).reset_index(drop=True)

# 🧠 뉴스 본문을 요약하는 함수
def summarize_news_content(url):
    try:
        res = requests.get(url, timeout=10)
        soup = BeautifulSoup(res.text, "html.parser")
        paragraphs = soup.find_all(['p', 'div'])

        full_text = ' '.join(p.get_text(strip=True) for p in paragraphs if len(p.get_text()) > 50)
        full_text = full_text.replace('\xa0', ' ').replace('\n', ' ').strip()

        if not full_text:
            return "❌ 뉴스 본문을 가져올 수 없습니다."

        prompt = f"다음 기사를 한국어로 300자 이내로 요약해줘:\n\n{full_text[:3000]}"
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=prompt
        )
        return response.text
    except Exception as e:
        return f"⚠️ 에러 발생: {e}"

# 🎨 Gradio UI
with gr.Blocks() as demo:
    gr.Markdown("## 📰 Fintech 뉴스 요약 챗봇")

    keyword = gr.Textbox(label="검색 키워드", placeholder="예: 핀테크")
    search_btn = gr.Button("🔍 뉴스 검색")

    news_dropdown = gr.Dropdown(label="뉴스 제목 선택", choices=[], interactive=True)
    summary_output = gr.Textbox(label="300자 뉴스 요약", lines=10)

    news_data_state = gr.State()

    # 🔄 뉴스 검색 및 드롭다운 갱신
    def update_news_dropdown(q):
        df = search_news(q)
        choices = df["title_clean"].tolist()
        default_value = choices[0] if choices else None
        return gr.update(choices=choices, value=default_value), df

    # ▶ 뉴스 선택 시 링크에서 요약 추출
    def summarize_selected_news(title, df):
        if df is None or df.empty:
            return "❌ 뉴스 데이터가 없습니다."
        try:
            link = df[df['title_clean'] == title]['originallink'].values[0]
            return summarize_news_content(link)
        except IndexError:
            return "❌ 링크를 찾을 수 없습니다."

    # 이벤트 연결
    search_btn.click(update_news_dropdown, inputs=keyword, outputs=[news_dropdown, news_data_state])
    news_dropdown.change(summarize_selected_news, inputs=[news_dropdown, news_data_state], outputs=summary_output)

demo.launch()
```
